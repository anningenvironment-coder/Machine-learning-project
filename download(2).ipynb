```python
import pandas as pd
import os

DATA_DIR = "/content/ecustfd"

annotations = pd.read_csv(os.path.join(DATA_DIR, "annotations.csv"))
portions = pd.read_csv(os.path.join(DATA_DIR, "portions.csv"))

# 去掉硬币，只保留食物
food_annotations = annotations[
    annotations["name"].str.lower() != "coin"
].copy()

# 统一食物名称的大小写，去除多余空格
food_annotations["name_key"] = (
    food_annotations["name"].str.strip().str.lower()
)
portions["type_key"] = (
    portions["type"].str.strip().str.lower()
)

# 同时匹配食物 ID 和食物类型
merged = food_annotations.merge(
    portions,
    left_on=["object_id", "name_key"],
    right_on=["id", "type_key"],
    how="left",
    indicator=True,
    validate="many_to_one"
)

print("食物标注行数：", len(food_annotations))
print("匹配后行数：", len(merged))
print("成功匹配：", (merged["_merge"] == "both").sum())
print("未匹配：", (merged["_merge"] == "left_only").sum())

print("\n匹配结果示例：")
display(
    merged[
        ["file", "object_id", "name", "weight_g", "calories", "_merge"]
    ].head(15)
)
```

```python
 # 1. 只保留匹配成功的食物标注
matched = merged[merged["_merge"] == "both"].copy()

# 2. 按图片文件名，把同一张图片中所有食物的热量相加
image_labels = (
    matched.groupby("file", as_index=False)
    .agg(
        total_calories=("calories", "sum"),
        object_ids=("object_id", lambda x: ",".join(sorted(set(x))))
    )
)

# 3. 排除存在未匹配食物标注的图片
bad_files = set(unmatched["file"])

image_labels = image_labels[
    ~image_labels["file"].isin(bad_files)
].copy()

print("可用于训练的图片数量：", len(image_labels))
print("被排除的异常图片数量：", len(bad_files))

print("\n标签示例：")
display(image_labels.head(10))

print("\n总热量统计：")
display(image_labels["total_calories"].describe())
```

```python
from sklearn.model_selection import GroupShuffleSplit

# 每个 object_ids 代表一份食物或一组食物的标识
splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.2,
    random_state=42
)

train_idx, val_idx = next(
    splitter.split(
        image_labels,
        groups=image_labels["object_ids"]
    )
)

train_df = image_labels.iloc[train_idx].reset_index(drop=True)
val_df = image_labels.iloc[val_idx].reset_index(drop=True)

print("训练集图片数量：", len(train_df))
print("验证集图片数量：", len(val_df))

# 检查两个集合是否包含相同的食物标识
train_groups = set(train_df["object_ids"])
val_groups = set(val_df["object_ids"])

print("训练集与验证集重叠的标识数量：", len(train_groups & val_groups))

display(train_df.head())
display(val_df.head())
```

```python
import os
import torch
from torch.utils.data import Dataset, DataLoader
from PIL import Image
from torchvision import transforms

DATA_DIR = "/content/ecustfd"
IMAGE_DIR = os.path.join(DATA_DIR, "images")

# 将所有图片调整到 224x224，并转换为 Tensor
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
])

class FoodCalorieDataset(Dataset):
    def __init__(self, dataframe, image_dir, transform=None):
        self.dataframe = dataframe.reset_index(drop=True)
        self.image_dir = image_dir
        self.transform = transform

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, idx):
        row = self.dataframe.iloc[idx]
        image_path = os.path.join(self.image_dir, row["file"])

        image = Image.open(image_path).convert("RGB")
        if self.transform:
            image = self.transform(image)

        calories = torch.tensor(
            [float(row["total_calories"])],
            dtype=torch.float32
        )

        return image, calories

train_dataset = FoodCalorieDataset(
    train_df, IMAGE_DIR, transform
)
val_dataset = FoodCalorieDataset(
    val_df, IMAGE_DIR, transform
)

train_loader = DataLoader(
    train_dataset, batch_size=32, shuffle=True
)
val_loader = DataLoader(
    val_dataset, batch_size=32, shuffle=False
)

# 先检查一个 batch
images, calories = next(iter(train_loader))

print("图片 Tensor 形状：", images.shape)
print("热量标签形状：", calories.shape)
print("前 5 个热量标签：", calories[:5].squeeze().tolist())
print("图片像素值范围：", images.min().item(), "到", images.max().item())
```

```python
import torch
import torch.nn as nn

class SimpleCNN(nn.Module):
    def __init__(self):
        super().__init__()

        self.features = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.ReLU(),

            nn.AdaptiveAvgPool2d((1, 1))
        )

        self.regressor = nn.Linear(128, 1)

    def forward(self, x):
        x = self.features(x)
        x = torch.flatten(x, start_dim=1)
        return self.regressor(x)
```

```python
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

model = SimpleCNN().to(device)

criterion = nn.MSELoss()
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)

print("使用设备：", device)
```

```python
def train_one_epoch(model, train_loader, criterion, optimizer, device):
    model.train()
    total_loss = 0.0
    total_samples = 0

    for images, calories in train_loader:
        images = images.to(device)
        calories = calories.to(device)

        optimizer.zero_grad()

        predictions = model(images)
        loss = criterion(predictions, calories)

        loss.backward()
        optimizer.step()

        batch_size = images.size(0)
        total_loss += loss.item() * batch_size
        total_samples += batch_size

    return total_loss / total_samples


def evaluate(model, val_loader, device):
    model.eval()

    total_squared_error = 0.0
    total_absolute_error = 0.0
    total_samples = 0

    with torch.no_grad():
        for images, calories in val_loader:
            images = images.to(device)
            calories = calories.to(device)

            predictions = model(images)

            total_squared_error += (
                (predictions - calories) ** 2
            ).sum().item()

            total_absolute_error += (
                (predictions - calories).abs()
            ).sum().item()

            total_samples += images.size(0)

    mse = total_squared_error / total_samples
    mae = total_absolute_error / total_samples
    rmse = mse ** 0.5

    return {"MAE": mae, "RMSE": rmse, "MSE": mse}
```

```python
import pandas as pd
from collections import defaultdict
from sklearn.model_selection import GroupShuffleSplit

# image_labels 是你之前已经生成的图片标签表
df = image_labels.reset_index(drop=True).copy()

# 将每张图片的食物 ID 拆成列表
df["id_list"] = df["object_ids"].fillna("").apply(
    lambda s: [x.strip() for x in s.split(",") if x.strip()]
)

# Union-Find：把共享同一份食物 ID 的图片归为一组
parent = list(range(len(df)))

def find(x):
    while parent[x] != x:
        parent[x] = parent[parent[x]]
        x = parent[x]
    return x

def union(a, b):
    ra, rb = find(a), find(b)
    if ra != rb:
        parent[ra] = rb

first_image_for_id = {}

for i, ids in enumerate(df["id_list"]):
    for food_id in ids:
        if food_id in first_image_for_id:
            union(i, first_image_for_id[food_id])
        else:
            first_image_for_id[food_id] = i

df["group_id"] = [find(i) for i in range(len(df))]

print("图片总数：", len(df))
print("独立分组数：", df["group_id"].nunique())
print("最大分组图片数：", df["group_id"].value_counts().max())
print("各分组大小前 10 名：")
print(df["group_id"].value_counts().head(10).tolist())
```

```python
from torch.utils.data import DataLoader

train_dataset = FoodCalorieDataset(train_df, IMAGE_DIR, transform)
val_dataset = FoodCalorieDataset(val_df, IMAGE_DIR, transform)
test_dataset = FoodCalorieDataset(test_df, IMAGE_DIR, transform)

train_loader = DataLoader(
    train_dataset, batch_size=32, shuffle=True
)
val_loader = DataLoader(
    val_dataset, batch_size=32, shuffle=False
)
test_loader = DataLoader(
    test_dataset, batch_size=32, shuffle=False
)

print("Train batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Test batches:", len(test_loader))
```

```python
import torch
import torch.nn as nn

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

model = SimpleCNN().to(device)
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

EPOCHS = 10
best_val_mae = float("inf")

for epoch in range(EPOCHS):
    train_mse = train_one_epoch(
        model, train_loader, criterion, optimizer, device
    )

    val_metrics = evaluate(model, val_loader, device)

    print(
        f"Epoch [{epoch+1}/{EPOCHS}] "
        f"Train MSE: {train_mse:.2f} | "
        f"Val MAE: {val_metrics['MAE']:.2f} | "
        f"Val RMSE: {val_metrics['RMSE']:.2f}"
    )

    if val_metrics["MAE"] < best_val_mae:
        best_val_mae = val_metrics["MAE"]
        torch.save(model.state_dict(), "best_simplecnn.pth")
        print("  Saved best model.")
```
